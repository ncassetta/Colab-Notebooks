# **Le funzioni**

````python
max(1, 5, 7, 2)
````

````python
min(5, -1, 30, - 10, 4, 7)
````

Le funzioni sono blocchi di codice riutilizzabili che eseguono un compito specifico. Immagina di avere un compito che devi fare più volte, come calcolare la media di un gruppo di numeri. Invece di scrivere lo stesso codice ogni volta, puoi creare una funzione che lo fa per te. Una volta definita, puoi "chiamare" o "invocare" la funzione ogni volta che ne hai bisogno.

**Perché sono utili le funzioni?**

1.  **Riutilizzabilità:** Puoi scrivere il codice una volta e usarlo molte volte.
2.  **Organizzazione:** Aiutano a organizzare il codice in blocchi logici, rendendolo più facile da leggere e capire.
3.  **Mantenibilità:** Se devi fare un cambiamento, lo fai solo in un posto (nella funzione), invece che in più punti del codice.
4.  **Astrazione:** Ti permettono di concentrarti su "cosa" fa un pezzo di codice, senza preoccuparti troppo di "come" lo fa internamente, una volta che sai che funziona.

**Come funzionano (concettualmente):**

*   **Input (Argomenti):** Le funzioni possono prendere degli input, chiamati argomenti, che sono i dati su cui devono lavorare. Ad esempio, per una funzione che calcola la media, gli argomenti sarebbero i numeri di cui vuoi calcolare la media.
*   **Elaborazione:** La funzione esegue le istruzioni al suo interno usando questi input.
*   **Output (Valore di ritorno):** Spesso, dopo aver fatto il suo lavoro, una funzione restituisce un risultato, che puoi poi usare nel resto del tuo codice.

Negli esempi che hai visto, `max()` e `min()` sono funzioni integrate in Python. `len()` è un'altra funzione che calcola la lunghezza di una stringa o di una sequenza.

La sintassi generale per chiamare una funzione in Python è:

```python
nome_funzione(argomento1, argomento2, ...)
```

**Ecco i punti chiave:**

1.  **`nome_funzione`**: È il nome della funzione che vuoi utilizzare (ad esempio, `max`, `min`, `len`).
2.  **Parentesi `()`**: Dopo il nome della funzione, devi sempre mettere le parentesi. È così che Python sa che stai cercando di *chiamare* la funzione, e non di riferirti ad essa come una variabile.
3.  **`argomento1, argomento2, ...` (gli argomenti)**: All'interno delle parentesi, passi i valori o le variabili su cui la funzione deve operare. Questi sono chiamati **argomenti**. Se ci sono più argomenti, li separi con una virgola. Ad esempio:
    *   `max(1, 5, 7, 2)`: `1`, `5`, `7`, `2` sono gli argomenti passati alla funzione `max`.
    *   `len("Giuseppina e Pippo")`: `"Giuseppina e Pippo"` è l'unico argomento passato alla funzione `len`.

**Casi speciali:**

*   **Funzioni senza argomenti:** Alcune funzioni non richiedono alcun argomento. In questi casi, le parentesi sono comunque obbligatorie, ma rimangono vuote, ad esempio `funzione_senza_argomenti()`.
*   **Il valore di ritorno:** Quando una funzione viene eseguita, spesso produce un risultato. Questo risultato viene chiamato "valore di ritorno". Puoi usare questo valore di ritorno direttamente in un'espressione, assegnarlo a una variabile o visualizzarlo, come hai fatto negli esempi.

In sintesi, per usare una funzione, scrivi il suo nome seguito da parentesi, e metti gli input (se necessari) all'interno delle parentesi.

````python
max("a", "c", "z", "e")
````

## La funzione print()

La funzione `print()` è una delle funzioni più fondamentali in Python. Il suo scopo principale è **visualizzare output sulla console** (o sullo schermo).

In pratica, qualunque cosa tu le passi come argomento, `print()` la convertirà in una stringa di testo e la mostrerà.

Ad esempio, se scrivi `print("Ciao mondo!")`, sullo schermo apparirà la frase "Ciao mondo!". Se le passi una variabile o un numero, `print()` mostrerà il loro valore.

**Argomenti della funzione `print()`:**

La funzione `print()` è molto versatile e può accettare **zero o più argomenti posizionali**, oltre a diversi argomenti chiave (keyword arguments) che ne modificano il comportamento.

*   **Argomenti posizionali:** Puoi passarle più elementi separati da virgole, e `print()` li stamperà tutti, separandoli di default con uno spazio.

    Esempio:
    ```python
    nome = "Alice"
    eta = 30
    print("Il mio nome è", nome, "e ho", eta, "anni.")
    # Output: Il mio nome è Alice e ho 30 anni.
    ```

*   **Argomenti chiave (Keyword Arguments):** Due dei più comuni sono:
    *   `sep` (separator): Specifica la stringa da usare per separare gli argomenti posizionali. Il default è uno spazio (`' '`).
    *   `end` (end of line): Specifica la stringa da aggiungere alla fine dell'output. Il default è un carattere di nuova riga (`'\n'`).

    Esempi:
    ```python
    print("Uno", "Due", "Tre", sep="-")
    # Output: Uno-Due-Tre

    print("Ciao", end=" ")
    print("Mondo!")
    # Output: Ciao Mondo!
    ```

## La funzione input()

La funzione `input()` è usata per **ottenere un input testuale dall'utente** tramite la console. Quando `input()` viene chiamata, il programma si ferma e attende che l'utente scriva qualcosa e prema 'Invio'. Ciò che l'utente digita viene letto dalla funzione come una stringa.

**Sintassi:**

```python
input(prompt)
```

**Argomenti della funzione `input()`:**

*   **`prompt` (opzionale):** Questo è l'unico argomento che `input()` accetta. È una stringa che viene visualizzata all'utente prima di attendere l'input. Serve per "chiedere" all'utente cosa deve inserire.

    *   Se `prompt` è fornito, verrà stampato sulla console (senza un carattere di nuova riga finale) e il cursore rimarrà sulla stessa riga, in attesa dell'input dell'utente.
    *   Se `prompt` non è fornito, il programma attenderà l'input dell'utente senza visualizzare alcun messaggio.

**Valore di ritorno:**

La funzione `input()` restituisce sempre una **stringa** contenente ciò che l'utente ha digitato. Anche se l'utente digita un numero, `input()` lo restituirà come una stringa. Se hai bisogno di usare l'input come un numero, dovrai convertirlo esplicitamente (ad esempio, con `int()` o `float()`).

**Esempio:**

```python
# Esempio con prompt
nome = input("Qual è il tuo nome? ")
print("Ciao, " + nome + "!")

# Esempio senza prompt
citta = input()
print("La tua città è: " + citta)

# Esempio con conversione a numero
eta_str = input("Quanti anni hai? ")
eta_int = int(eta_str)
print("Tra un anno avrai", eta_int + 1, "anni.")
```

````python
nome = input("Come ti chiami? ")
print("Ciao,", nome)
citta = input("Dove abiti? ")
print("La tua città è", citta)
````

## **Le f-string (Formatted String Literals)**

Le f-string sono un modo potente e leggibile per creare stringhe formattate in Python, introdotte a partire dalla versione 3.6. Ti permettono di incorporare espressioni Python all'interno di stringhe letterali, rendendo la formattazione del testo molto più semplice e intuitiva rispetto ai metodi precedenti (come il metodo `format()` o l'operatore `%`).

**Come funzionano:**

Una f-string è una stringa normale preceduta dalla lettera `f` (o `F`). All'interno della stringa, puoi inserire espressioni Python racchiuse tra parentesi graffe `{}`. Python valuterà queste espressioni e inserirà il loro risultato direttamente nella stringa.

**Sintassi di base:**

```python
f"Il tuo testo {espressione_python} altro testo {altra_espressione}"
```

**Vantaggi principali:**

1.  **Leggibilità:** Sono molto più facili da leggere e scrivere, specialmente quando si incorporano molte variabili o espressioni.
2.  **Concise:** Il codice è più corto e diretto.
3.  **Velocità:** Sono generalmente più veloci di altri metodi di formattazione delle stringhe.

**Esempi:**

```python
nome = "Anna"
eta = 30

# Senza f-string (meno leggibile)
print("Ciao " + nome + ", hai " + str(eta) + " anni.")
print("Ciao {}, hai {} anni.".format(nome, eta))

# Con f-string (più leggibile e conciso)
print(f"Ciao {nome}, hai {eta} anni.")

# Puoi inserire direttamente espressioni e chiamate a funzioni
prezzo = 19.99
quantita = 3
totale = prezzo * quantita
print(f"Hai acquistato {quantita} articoli per un totale di {totale:.2f} euro.") # Il :.2f formatta a due cifre decimali

# Anche operazioni matematiche o chiamate a funzioni
import math
raggio = 5
area_cerchio = math.pi * raggio**2
print(f"L'area di un cerchio con raggio {raggio} è {area_cerchio:.2f}.")
```

In sintesi, le f-string sono lo strumento preferito per la formattazione delle stringhe in Python moderno grazie alla loro chiarezza, concisione e potenza.
