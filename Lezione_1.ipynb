## Espressioni con stringhe


### Cosa sono le stringhe e quali operatori si possono usare?

Una **stringa** è una sequenza di caratteri. Puoi immaginarla come una 'parola' o una 'frase' racchiusa tra virgolette (singole `' '` o doppie `" "`). Per esempio, `'ciao'` o `"Python è divertente"` sono stringhe.

In Python, puoi usare due operatori principali con le stringhe:

1.  **Concatenazione (`+`)**: Questo operatore unisce due o più stringhe in una singola stringa. Ad esempio:
    ```python
    "Ciao" + " " + "mondo!"
    # Output: 'Ciao mondo!'
    ```

2.  **Moltiplicazione per un intero (`*`)**: Questo operatore ripete la stringa un certo numero di volte. Ad esempio:
    ```python
    3 * "abc"
    # Output: 'abcabcabc'
    ```

**Attenzione:** Come hai visto con l'errore `TypeError` nella cella precedente, non puoi "sommare" direttamente un numero intero (`int`) e una stringa (`str`) usando l'operatore `+`. Python richiede che entrambi gli operandi siano dello stesso tipo per l'operazione di concatenazione (`+`) o che il numero sia un intero per l'operazione di moltiplicazione (`*`) con una stringa.

### Cosa sono le espressioni booleane?

Un'**espressione booleana** è un'espressione che può avere solo due possibili risultati: **`True`** (vero) o **`False`** (falso). Immagina di fare una domanda a Python a cui può rispondere solo con un sì o un no.

Questi due valori, `True` e `False`, sono costanti predefinite in Python e sono molto importanti per prendere decisioni nel codice. Per questo motivo, vanno scritti **senza virgolette e sempre con l'iniziale maiuscola**.

Hai già usato diversi **operatori di confronto** (o relazionali) per creare espressioni booleane nelle celle precedenti, come:

*   `>` (maggiore di)
*   `<` (minore di)
*   `>=` (maggiore o uguale a)
*   `<=` (minore o uguale a)
*   `==` (uguale a) - *Attenzione: non confonderlo con `=` che serve per assegnare un valore!*
*   `!=` (diverso da)

Inoltre, esistono gli **operatori logici** che permettono di combinare più espressioni booleane:

*   **`and`**: Restituisce `True` se *entrambe* le espressioni booleane sono `True`. Altrimenti, restituisce `False`.
    *Esempio: `7 > 6 and 5 > 6` (True e False -> False)*

*   **`or`**: Restituisce `True` se *almeno una* delle espressioni booleane è `True`. Restituisce `False` solo se *entrambe* sono `False`.
    *Esempio: `4 > 6 or 5 > 6` (False o False -> False)*

*   **`not`**: Inverte il valore booleano di un'espressione. Se è `True` diventa `False`, e se è `False` diventa `True`.
    *Esempio: `not (5 > 1)` (not True -> False)*

````python
5 > 1
````

````python
5 >= 5
````

````python
5 != 7
````
