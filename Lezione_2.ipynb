## Regole per i Nomi delle Variabili in Python

Per assegnare un nome alle tue variabili in Python, ci sono alcune regole importanti da seguire:

1.  **Caratteri Validi**: I nomi delle variabili possono contenere lettere (maiuscole e minuscole), numeri e il carattere underscore (`_`).
    *   `nome_valido`
    *   `var123`
    *   `_mia_variabile`

2.  **Iniziare con Lettera o Underscore**: Un nome di variabile **non può** iniziare con un numero. Deve iniziare con una lettera o un underscore.
    *   `123variabile` (Non valido)
    *   `miaVariabile` (Valido)

3.  **Maiuscole e Minuscole Distinte (Case-Sensitive)**: Python distingue tra maiuscole e minuscole. Questo significa che `miaVariabile`, `miavariabile` e `MiaVariabile` sono considerate tre variabili diverse.

4.  **Parole Chiave Riservate**: Non puoi usare le parole chiave (keywords) di Python come nomi di variabili. Queste parole hanno un significato speciale nel linguaggio (ad esempio `if`, `else`, `while`, `for`, `print`, `True`, `False`, `None`, ecc.). Se provi a usarle, Python ti darà un errore.

5.  **Nomi Descrittivi**: È una buona pratica dare nomi significativi alle variabili che indichino lo scopo o il contenuto della variabile. Questo rende il tuo codice più facile da leggere e capire.
    *   `età_utente` è meglio di `eu`

6.  **Convenzioni di Stile (PEP 8)**: Sebbene non siano regole obbligatorie, esistono delle convenzioni per i nomi delle variabili (standard PEP 8) che rendono il codice Python più uniforme e leggibile:
    *   Usa il `snake_case` (tutto minuscolo con underscore per separare le parole) per i nomi delle variabili e delle funzioni, ad esempio `velocita_massima`.
    *   Usa `CAPS_LOCK` per le costanti, ad esempio `GRAVITA`.

Seguendo queste regole e convenzioni, scriverai un codice Python più pulito e professionale!

# **Commenti in Python**

I commenti sono porzioni di testo all'interno del codice che vengono ignorate dall'interprete Python. Sono utili per spiegare il codice, renderlo più leggibile e per scopi di documentazione.

### Commenti a riga singola

Per un commento che occupa una singola riga, si usa il simbolo del cancelletto (`#`). Tutto ciò che segue `#` sulla stessa riga viene considerato un commento.

Esempio:
```python
# Questo è un commento a riga singola
x = 10  # Anche questa parte della riga è un commento
```

**Ricorda:** Usare i commenti in modo efficace rende il tuo codice molto più comprensibile sia per te stesso in futuro che per altri che potrebbero leggerlo!
