## **L'istruzione if**

Un'istruzione `if` viene usata per eseguire un blocco di codice solo se una certa condizione è vera.

Nel codice sopra:
*   `if n > 10:` controlla se il valore di `n` è maggiore di 10.
*   Se la condizione `n > 10` è vera, la riga indentata `print("Bravo!")` viene eseguita.
*   Se la condizione è falsa (come nell'esempio in cui `n` era `7`), la riga indentata viene saltata e il programma prosegue con la riga successiva non indentata, cioè `print("Ciao!")`.

```python
n = int(input("Scrivi un numero > 10 "))
if n > 10:               # ATTENTi ai : finali
    print("Bravo!")      # INDENTATE la riga con 2 Tab (NON CON GLI SPAZI!)
else:
    print("Avevo detto > 10!")
print("Ciao!")           # Questa riga non è indentata
```

## **L'istruzione else**

Nel codice sopra, se `n` non è maggiore di 10 (cioè, `n <= 10`), il blocco di codice sotto `else:` viene eseguito, stampando "Avevo detto > 10!". In questo modo, il programma può reagire in modo diverso a seconda che la condizione sia vera o falsa.

## Esempio
Nel ristorante-pizzeria "da Peppe" abbiamo i seguenti prezzi:

*   Pizza margherita 7.50 €
*   Pizza napoletana 9.00 €
*   Pasta alla carbonara 12.00 €
*   Pasta all'amatriciana 12.50 €

Scrivere un programma che chieda all'utente: "Vuoi la pasta o la pizza?"

Se l'utente risponde "pasta" deve chiedere "carbonara o amatriciana?"

Se risponde "pizza" deve chiedere "margherita o napoletana?"

Alla fine deve stampare il prezzo: "Costo: xx €"



```python
# Stampa una riga di benvenuto
print("Benvenuto al ristorante-pizzeria 'da Peppe'!")
# Prima domanda: pasta o pizza?
scelta_piatto = input("Vuoi la pasta o la pizza? ")
# Se l'utente ha scelto "pizza" continua qui
if scelta_piatto == "pizza":
    # Seconda domanda per la pizza
    scelta_tipo_pizza = input("margherita o napoletana? ")
    if scelta_tipo_pizza == "margherita":
        prezzo = 7.50
    elif scelta_tipo_pizza == "napoletana":
        prezzo = 9.00
# Se l'utente ha scelto "pasta" continua qui
elif scelta_piatto == "pasta":
    # Seconda domanda per la pasta
    scelta_tipo_pasta = input("carbonara o amatriciana? ").lower()
    if scelta_tipo_pasta == "carbonara":
        prezzo = 12.00
    elif scelta_tipo_pasta == "amatriciana":
        prezzo = 12.50
# Non indentato: qui si riuniscono i vari rami
print(f"Costo: {prezzo:.2f} €")
```
